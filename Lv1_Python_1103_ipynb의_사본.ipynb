```python
# 복소수 생성
c = 3 + 4j
print(c)         # 출력: (3+4j)

# 실수부와 허수부 acess
print(c.real)    # 출력: 3.0
print(c.imag)    # 출력: 4.0

# real과 imag는 메서드가 아니라 속성이라서 뒤에 ()를 붙이지 않는다
# 복소수의 real, imag 속성은 실수형인 float 값으로 제공된다. 3이 아니라 3.0으로 출력
```

```python
import pandas as pd

# 데이터프레임
df = pd.DataFrame({'name': ['Alice', 'Bob'], 'age': [25, 30]})


# { } (중괄호): 딕셔너리, 표의 '열 이름(Column)' 세팅
# [ ] (대괄호): 리스트", 그 열에 들어갈 '실제 세로 데이터(Value)' 세팅
# : (콜론): 열 이름과 실제 데이터를 '연결'하는 매핑 기호

##  1단계: { } 중괄호 (비유적으로) 엑셀 표의 컬럼 이름표
# 파이썬의 딕셔너리 {'Key': 'Value'} 구조
# 판다스는 이 구조를 그대로 빌려와서 Key를 표의 '열 이름(Column Name)'으로 사용
  # 'name': ... ->  'name'이라는 열 생성...
  # 'age': ...  ->  'age'이라는 열 생성

## 2단계: [ ] 대괄호 "입력할 데이터"
# 이름표 아래 데이터 채우기
# 이때 여러 개의 데이터를 순서대로 묶어주기 위해 파이썬의 기본 리스트([]) 사용
#  ['Alice', 'Bob'] -> name 열의 첫 번째 칸에는 Alice, 두 번째 칸에는 Bob을 입력
#  [25, 30] -> "age 열의 첫 번째 칸에는 25, 두 번째 칸에는 30 입력

## 3단계: pd.DataFrame(...) : 위 데이터들을 묶어 엑셀 표 형식으로 변환
```

```python
scores = [85, 50, 70, 40, 95]
pass_count = 0  # 합격자 수, 처음에는 0으로 설정

for score in scores:
    if score >= 60:
        pass_count += 1  # 합격자 발생 시 1 증가

print(f"합격자 수: {pass_count}명")
```

```python
for number in range(1, 6):
    if number == 3:
        break
    print(number)

# 1 이상, 6미만 숫자인 1, 2, 3, 4, 5 중 숫자 3을 반환
# 숫자 3부터 반복문 종료
```

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)

# 숫자 3일시 3출력 않고 다음 숫자인 4로 넘어가 반환 진행
```

```python
# for문 안에 for문이 또 있는 중첩 반복문
for dan in range(2, 5):           # 2단, 3단, 4단
    print(f"[{dan}단]")

    for number in range(1, 10):  # 각 단에서 1~9 반복
        print(f"{dan} × {number} = {dan * number}")

    print()                      # 한 단이 끝나면 빈 줄 출력
```

```python
# f-string(포맷 스트링) 문법
# f : 포맷(Format, 서식)을 의미하는 영단어의 앞 글자

print(f"[{dan}단]")
```

```python
print("[{dan}단]")   # f 빼면...
                     # f를 붙여주면 {dan} 부분을 '변수'로 인식 -> 변수 안에 담긴 실제 값을 출력.
```

```python
# 비교
print("[" + str(dan) + "단]")
```

```python
# f-string 장점 : 중괄호 안에서 연산 가능
a = 5
b = 3
print(f"{a} + {b}의 결과는 {a + b}입니다.") # 출력: 5 + 3의 결과는 8입니다.
```

```python
# f-string 장점 : 소수점 자리수 조절

pi = 3.141592
print(f"원주율: {pi:.2f}") # 출력: 원주율: 3.14 (소수점 둘째 자리까지 제한)
```
