```python
import numpy as np # 숫자와 배열 계산
import pandas as pd # 표 형태의 데이터 처리

from sklearn.datasets import load_iris # 붓꽃 예제 데이터 불러오기
from sklearn.model_selection import train_test_split # 학습·테스트 데이터 분리
from sklearn.preprocessing import StandardScaler # 데이터 표준화
from sklearn.linear_model import LogisticRegression # 분류 모델 생성
from sklearn.metrics import accuracy_score, classification_report # 예측 정확도 계산, 분류 성능 상세 평가
```

image.png

```python
# 1. 데이터 수집 및 준비
# from sklearn.datasets import load_iris
# x to y

iris = load_iris()
X = iris.data # 입력 데이터
y = iris.target # 정답 데이터 (예측하려는 목표 변수)

```

```python
### 데이터 프레임 변환
# X(특성)를 데이터프레임으로 변환 (컬럼명 지정)
df = pd.DataFrame(X, columns=iris.feature_names)

# 3. y(타겟) 열 추가
df['target'] = y

# 결과 확인
df.tail(10) # 데이터프레임의 마지막 10개 행을 보여주는 코드

# Pandas 데이터프레임은 기본적으로 행 번호를 0부터 시작
# target 값에 2라고 뜨는건 명목형 변수(꽃 이름)를 의미
```

```python
# 2. 데이터 전처리
# 데이터를 훈련 세트와 테스트 세트로 분리
# from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# test_size=0.3	전체 데이터의 30%를 테스트용으로 사용
# random_state=42	무작위 분할 결과를 재현할 수 있도록 난수 시드 고정
```

```python
X_train
```

```python
y_train
```

```python
# 특성 스케일링
# from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train) # fit 훈련 데이터에서 각 변수의 평균과 표준편차를 계산
X_test = scaler.transform(X_test)

# 특성 스케일링은 변수들이 가진 값의 크기나 범위를 조정하는 전처리 작업
```

```python
# 3. 모델 선택 및 학습
# from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
model.coef_ # 실제 로지스틱 회귀 결과를 보려면
```

```python
model.intercept_ # 회귀계수뿐 아니라 절편도 보고 싶다면
```

```python
# 4. 모델 평가
# from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}") # .2f 소수점 둘째 자리까지 고정소수점 형식으로 표시
print("Classification Report:") # 클래스별 평가지표 필요
print(classification_report(y_test, y_pred))

# 전체 정확도뿐 아니라 특정 품종별 결과값도 봐야 좋은 코 설계?!
# 정밀도: 모델이 특정 품종이라고 예측한 것 중 실제로 그 품종이었던 비율
# 재현율: 실제로 특정 품종인 꽃 중 모델이 정확하게 찾아낸 비율
# f1 : 정밀도와 재현율을 함께 고려한 지표
# macro avg	모든 품종의 지표를 동일한 비중으로 단순 평균
# weighted avg	각 품종의 실제 데이터 개수(support)를 가중치로 평균
```

```python
# 하이퍼파라미터 튜닝
# GridSearchCV 또는 RandomizedSearchCV로 하이퍼파라미터 조정 가능

# GridSearchCV는 개발자가 지정한 하이퍼파라미터 조합을 모두 시험하고, 교차검증을 통해 좋은 조합을 찾는 도구
# RandomizedSearchCV는 모든 조합을 시험하지 않고, 지정한 횟수만큼 무작위로 설정값을 뽑아 검증하는 방법

# 파라미터 모델이 학습한 값, 하이퍼파라미터 학습 방법을 제어 하는 설
```

```python
# 5. 예측
# 예제 입력 데이터로 모델 예측
example_data = np.array([[20.0, 18.6, 5.4, 5.2]])
example_data_scaled = scaler.transform(example_data)
prediction = model.predict(example_data_scaled)
print(f"Predicted class: {iris.target_names[prediction]}")
```
